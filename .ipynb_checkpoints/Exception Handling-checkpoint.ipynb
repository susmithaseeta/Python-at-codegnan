### Exception Handling

- Exception is nothing but it is an error, whenever an section is occur it stops the program execution & it returns type of execption & followed by exception message
### Type of Exception 0r Error
    1. Complie time error / Syntax error : The syntax error raises during the complication time of the program. Those errors called it as syntax error / complier time error.
    Ex: Intendation error, Syntax error 
    2. Run time error : The runtime error raises during runtime of a program. Those errors called as Runtime Error.
    Ex: Key Error, Type Error, Division Error, Index Error, type Error.....
    3. Logical error
=> we can't handle syntax error, but we can handle the runtime error by using exception handling (i.e try-except) 

## Exception Handling Blocks:
syntax:                                                                    <br>
    try:                                                                   <br>
        # Block of code that might contain exception                       <br>
    except:                                                                <br>
        # Exception Handling Block                                         <br>
    else: [option]                                                         <br>
        # This block exception there is no exception raised in try block   <br>
    finally: [optional]                                                    <br>
        # This block always except every exception raised or not.          <br>

```python
# try block with exception 
try:
    list = [1,2,3]
    print(list[10])
    print("hi")
except:
    print("something wrong in try block")
else:
    print("No Exception raised in try block")
finally:
    print("Program done")
    
```

```python
# try block without exception
try:
    list = [1,2,3]
    print(list[2])
    print("hi")
except:
    print("something wrong in try block")
else:
    print("No Exception raised in try block")
finally:
    print("Program done")
```

```python
# we can't handle the syntax error 
try:
    print("Hello')
except:
    print("something wrong in try block")
else:
    print("No Exception raised in try block")
finally:
    print("Program done")
```

```python
try:
    a = 10
    b = 0
    c = a / b
    print(c) 
except ZeroDivisionError as e:
    print(f"Zero division not possible: {e}")
except ValueError as e:
    print(f"Value error raised in try block: {e}")
# except Exception as e:
#   print(f"Something wrong in try block: {e}")
```

```python
# It execute the specific exception block when specified exception raised otherwise,
#             it executes the common except block
try:
    a = 10
    b = a.append(1)
    c = a / b
    print(c) 
except ZeroDivisionError as e:
    print(f"Zero division not possible: {e}")
except ValueError as e:
    print(f"Value error raised in try block: {e}")
except Exception as e: # common exception block with Exception class
   print(f"Something wrong in addition.add: {e}")
```
