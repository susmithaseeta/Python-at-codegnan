- Methods: The function defination is implemented inside the class those methods called as Methods
    - Types Of Methods
     1. Instance Method
     2. Class Method
     3. static Method

```python
# class attribute - school
# instance attribute - id,name,class
class Student:
    school = "Codegnan school"
    def __init__(self, stdid, name,standard):
        self.name = name 
        self.stdid = stdid
        self.standard = standard
    def student_details(self):
        return f"My name is {self.name} and I am from {self.standard} class"
s = Student(429,"Susmitha",10)
print(s.name)
print(s.school)
print(s.student_details())
```

- Instance Method : The method defination takes forst parameter as 'self' thoes methods called as `Instance Method`. This methods works on instance attributes.   <br>
 `Syntax:`<br>
    `        class class_name:`                  <br>
    `           def method_name(self):`               <br>
    `             #blockcode`                       <br>
  
- Class Method : The method defination starts with `classmethod` decorator and it takes first parameter as `cls` those methods called as `Class Methods`. This methods works on class Attribute.<br>
  `Syntax:`<br>
    `        class class_name:`                  <br>
    `            @classmethod `                  <br>
    `            def method_name(cls, parameters):`    <br>
    `                 #blockcode`                       <br>

  
- Static Method : The method defination starts with `staticmethod` decorator and it doesn't take first parameter as `self or cls` those methods called as `Static Methods`. This methods works on except instance attributes and class Attribute.<br>
  `Syntax:`<br>
    `        class class_name:`                  <br>
    `            @classmethod `                  <br>
    `            def method_name(parameters):`         <br>
    `                 #blockcode`                       <br>

- Initialization Constructor (__init__) : The initialization constructor is used for to initialize the instance attributes data and this constructor takes first parameter as self and this constructor executes whenever an object is created for that class.

```python
# Instance Method:
# class attribute - school
# instance attribute - id,name,class
class Student:
    school = "Codegnan school" # class attribute
    def __init__(self, stdid, name,standard):
        self.name = name # instance attribute
        self.stdid = stdid # instance attribute
        self.standard = standard # instance attribute
    def student_details(self): # instance Method
        return f"My name is {self.name} and I am from {self.standard} class"
    # class method
    @classmethod
    def update_school(cls, new_school_name):
        cls.school = new_school_name
        return "School name Updated"
    @staticmethod
    def percentage(marks : list):
        return (sum(marks)/300) * 100
        
s1 = Student(429,"Susmitha", 9)
s2 = Student(430,"Akshya", 10)
print(s1.school, s2.school)
print(s.student_details())
print(s1.update_school("CODEGNAN school"))
print(s1.school, s2.school)
print(s2.percentage([75,85,95]))
```

#### Encapsulation:
    - Encapsulation means, wraping the methods and attributes in single unit that one is class. The main
aim of Encapsulation is to restrict the direct access of attributes and methods.
    - This type of Restriction can implement by using `Access modifiers`.
    - Types of Access Modifier
        1. Public
        2. Protected
        3. Private
    - In Python we don't have any keywords to represent it public, protected and private, but by using 
underscore(_) can define.
    1. Public : The Public Members(i.e attributes + method) doesn't start with underscore(_). This public members can access any where in the program.
    2. Protected : The Protected Members(i.e attributes + method) starts with single underscore(_). This members can access in inside the class and in subclass also.
    3. Private : The Private Members(i.e attributes + method) start with double underscore(__). This members can access in only inside.

```python
class Bank:
    def __init__(self, name, account,  password, balance):
        self.name = name
        self.account = account
        self.__password = password
        self._balance = balance
    def __total_users(self): # getter method
        return "Total users : 101"
    def get_balance(self): # getter method
        return f"Current Balance is:{self._balance}"
    def deposite(self, deposite_amount): # setter method
        self._balance += deposite_amount
        return f"{deposite_amount} Deposited Successfully"
        
u1 = Bank("Susmitha", 1001, "Susmitha@123", 10000)
print(u1.name)
print(u1.get_balance())
print(u1.deposite(5000))
print(u1.get_balance())
```
