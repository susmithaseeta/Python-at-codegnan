### Polymorphism

```python
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and I am {self.age} years old" 

class Student(Person):
    def __init__(self, stdid, name, age, gender, branch):
        super().__init__(name, age, gender)
        self.stdin = stdid
        self.branch = branch
    def details(self):
        return f"My name is {self.name} and I am belongs to {self.branch} Branch"
s = Student(101, "Susmitha", 21, "F", "CSE")
print(s.name)
print(s.branch)
print(s.details())
```

```python
# Method overriding
# Even if you define different number of parameters in methods, always child method overrides parent class method
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self, location):
        return f"My name is {self.name} and I am staying in {location}" 

class Student(Person):
    def __init__(self, stdid, name, age, gender, branch):
        super().__init__(name, age, gender)
        self.stdin = stdid
        self.branch = branch
    def details(self):
        print(super().details())
        return f"My name is {self.name} and I am belongs to {self.branch} Branch"
s = Student(101, "Susmitha", 21, "F", "CSE")
print(s.name)
print(s.branch)
print(s.details("Indian"))
```

```python
# Method overriding
# Even if you define different number of parameters in methods, always child method overrides parent class method
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def details(self, location):
        return f"My name is {self.name} and I am staying in {location}"


class Student(Person):
    def __init__(self, name, age, gender, branch, rollno):
        super().__init__(name, age, gender)
        self.branch = branch
        self.rollno = rollno

    def details(self):
        print(super().details("Hyderabad"))
        return f"My name is {self.name} and I belong to {self.branch} branch."


s = Student("Susmitha", 21, "Female", "CSE", 101)
print(s.details())
```

```python
# Method Overloading
class Math:
    def add(self, a, b):
        return a + b
    def add(self, a, b, c):
        return a + b + c
    def add(self, a, b, c, d):
        return a + b + c + d
m = Math()
print(m.add(1,2,3,4))
# print(m.add(1,2,3))
# print(m.add(1,2))
```

```python
# Method Overloading
# Directly python doesn't support method overloading but, we can achive same functional
#          through the default argumentd or Arbitary Arguments.
class Math:
    def add(self, a, b, c = 0, d = 0):
        return a + b + c + d
m = Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
# Magic methods or Special Methods
class point:
    def __init__(self, x):
        self.x = x
p1 = point(5)
print(p1)
```

```python
# __str__() magic methods
class point:
    def __init__(self,x):
        self.x = x
    def __str__(self): # String magic methods
        return f"This is point class and point values is {self.x}"
p1 = point(5)
print(p1)
```

```python
# __str__() magic methods
class point:
    def __init__(self,x):
        self.x = x
    def __str__(self): # String magic methods
        return f"This is point class and point values is {self.x}"
    def __add_(self, other):
        return self.x + other.x
p1 = point(5)
print(p1)
p2 = point(3)
# print(p1.x + p2.x)
print(p1 + p2)
```

```python
# __str__() magic methods
class point:
    def __init__(self,x):
        self.x = x
    def __str__(self): # String magic methods
        return f"This is point class and point values is {self.x}"
    def __add__(self, other):
        return point(self.x + other.x)
p1 = point(5)
print(p1)
p2 = point(3)
# print(p1.x + p2.x)
p3 = p1 + p2
print(type(p3))
print(p3)
```

```python
# __str__() magic methods
class point:
    def __init__(self,x):
        self.x = x
    def __str__(self): # String magic methods
        return f"This is point class and point values is {self.x}"
    def __add__(self, other):
        return self.x + other.x
    def __sub__(self, other):
        return self.x - other.x
    def __mul__(self, other):
        return self.x * other.x
    def __div__(self, other):
        return self.x / other.x
    def __floordiv__(self, other):
        return self.x // other.x
    def __mod__(self, other):
        return self.x % other.x
    def __pow__(self, other):
        return self.x ** other.x
p1 = point(5)
p2 = point(3)
p3 = p1 + p2
p4 = p1 - p2
p5 = p1 * p2
#p6 = p1 / p2
p7 = p1 // p2
p8 = p1 % p2
p9 = p1 ** p2
print(p3)
print(p4)
print(p5)
#print(p6)
print(p7)
print(p8)
print(p9)
```
