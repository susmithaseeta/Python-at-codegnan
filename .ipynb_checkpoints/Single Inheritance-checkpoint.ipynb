```python
# Types of Inheritance
# Single Inheritance
# Without child class implementation
# When child class don't have constructor, in that case, parent class constructor behave like child class constructor
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and I am {self.age} years old"
        
class Student(Person):
    pass
s = Student("Susmitha", 21, "Female")
print(s.name)
print(s.details())
```

```python
# Types of Inheritance
# Single Inheritance
# child class with constructor
# When child class don't have constructor, in that case, parent class constructor behave like child class constructor
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and I am {self.age} years old"
        
class Student(Person):
    def __init__(self, stdid, name, age, branch):
        self.stdid = stdid
        self.name = name
        self.age = age
        # self.gender = gender
        self.branch = branch
s = Student(101, "Susmitha", 21, "Python")
print(s.name)
print(s.branch)
# print(s.gender) # AttributeError: 'Student' object has no attribute 'gender'
print(s.details())
```

```python
# Types of Inheritance
# Single Inheritance
# child class with constructor
# When child class don't have constructor, in that case, parent class constructor behave like child class constructor
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and I am {self.age} years old and I am a {self.gender}"
        
class Student(Person):
    def __init__(self, stdid, name, age, branch):
        self.stdid = stdid
        self.name = name
        self.age = age
        # self.gender = gender
        self.branch = branch
s = Student(101, "Susmitha", 20, "Python")
print(s.name)
print(s.branch)
# print(s.gender) # AttributeError: 'Student' object has no attribute 'gender'
print(s.details()) # It raises AttributeError because Student reference doesn't contain gender attribute
```

```python
# Types of Inheritance
# 1. Single Inheritance
# child class with constructor
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and I am {self.age} years old and I am a {self.gender}"
        
class Student(Person):
    def __init__(self, stdid, name, gender, age, branch):
        super().__init__(name,age,gender)
        self.stdid = stdid
        self.branch = branch
        
s = Student(101, "Susmitha", "Female", 21, "Python")
print(s.name)
print(s.branch)
# print(s.gender) # AttributeError: 'Student' object has no attribute 'gender'
print(s.details())
```

```python
# 2. Multi level Inheritance 
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old"
    
class Student(Person):
    def __init__(self, name, age,branch):
        super().__init__(name, age)
        self.branch = branch
        
class Employee(Student):
    def __init__(self, name, age,branch, salary):
        super().__init__(name, age, branch)
        self.salary = salary
    def get_salary (self, Bonous = 0):
        self.salary += Bonous
        return self.salary
e = Employee("Susmitha", 21, "HYD", 50000)
print(e.name)
print(e.branch)
print(e.details())
print(e.get_salary())
print(e.get_salary(Bonous = 10000))
```

```python
# 3. Multiple Inheritance
class Father:
    def __init__(self,father_name):
        self.father_name = father_name

class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name

# child class
class Son(Father, Mother):
    def __init__(self, name, age, father_name, mother_name):
        Father.__init__(self, father_name)
        Mother.__init__(self, mother_name)
        self.name = name
        self.age = age
    pass
s = Son("Susmitha", 21, "Sangamesh", "Kalpana")
print(s.father_name)
print(s.mother_name)
print(s.name)
```

```python
# 3. Multiple Inheritance
class Father:
    def __init__(self,father_name):
        self.father_name = father_name
    def work(self):
        return "Father working as Engineer"

class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name
    def work(self):
        return "Mother working as Teacher"

# child class
class Son(Father, Mother):
    def __init__(self, name, age, father_name, mother_name):
        Father.__init__(self, father_name)
        Mother.__init__(self, mother_name)
        self.name = name
        self.age = age
    def parent_work(self):
        print(super().work())
        print(super().work())
        
s = Son("Susmitha", 21, "Sangamesh", "Kalpana")
# print(s.father_name)
# print(s.mother_name)
# print(s.name)
# print(Son.mro())
# print(s.work())
print(s.parent_work())
```

```python
# Hierachical Inheritance
# Parent Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def details(self):
        return f"Name: {self.name}, Age: {self.age}"


# Child Class 1
class Student(Person):
    def __init__(self, name, age, student_id, branch):
        super().__init__(name, age)
        self.student_id = student_id
        self.branch = branch

    def student_info(self):
        return f"Student ID: {self.student_id}, Branch: {self.branch}"


# Child Class 2
class Employee(Person):
    def __init__(self, name, age, emp_id, department):
        super().__init__(name, age)
        self.emp_id = emp_id
        self.department = department

    def employee_info(self):
        return f"Employee ID: {self.emp_id}, Department: {self.department}"


# Creating Student Object
s = Student("Susmitha", 21, 101, "Python")

print("Student Details")
print(s.details())
print(s.student_info())

print()

# Creating Employee Object
e = Employee("Amir", 28, 201, "HR")

print("Employee Details")
print(e.details())
print(e.employee_info())
```
