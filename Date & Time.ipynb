```python
from datetime import datetime
current_timestamp = datetime.now() # To get the current timestamp
print(current_timestamp)
today = current_timestamp.date() # To extract date from current timestamp
print(today)
curr_time = current_timestamp.time() # It extract the time from current timestamp
print(curr_time)

# Extract day, month, & year
day = today.day
month = today.month
year = today.year
print(day, month, year)

# Extract hour, mintue, & seconds from time
print("Hour:",curr_time.hour)
print("Minute:",curr_time.minute)
print("Second:",curr_time.second)

# Change date format
print(today.strftime("%d / %m / %y"))
print(today.strftime("%d - %a / %b / %Y"))
print(today.strftime("%I : %M : %p"))
print(curr_time.strftime("%I : %M : %p"))
print(curr_time.strftime("%H : %M : %S"))
```

```python
# Time implementation
import time
from IPython.display import clear_output
hour_counter = min_counter = second_counter = 0
while True:
    print(f"{hour_counter} : {min_counter} : {second_counter}")
    time.sleep(1) # It pause the execution for 1sec
    second_counter += 1
    if second_counter == 60:
        min_counter += 1
        second_counter = 0 # Reset seconds counter to 0
        if min_counter == 60:
            hour_counter += 1
            min_counter = 0 # Reset mintue counter to 0
            if hour_counter == 24:
                hour_counter = 0
    clear_output(wait = True)
```
