### File Handling

#### File:
- The file is used to stored data in electronic format in system.
#### File Operations:
##### 1. Opening a file : open()
       - To open a file by using open() in that we need to pass file name & file opening mode.
     Syntax:
        file_obj = open(file_name,mode)  
         # read / write operations      
         # close file

###### File Opening Mode:                   
        1. r : read - opening file in read mode , if file not exist it raises FileNot not error.
        2. w : write - openeing a file in write mode, if file not exist it creates new file otherwise it a raises a previous contant.
        3. a : append - opening a file in append mode, if file not exist it creates new file otherwise(file already exist) it add contant to the       previous contant. 
        4. x : create - create new file.
        5. b : binary - This mode is used to perform operation on binary files for writing `wb` mode for reading `rb` mode.
        6. r+ : read + write
        7. w+ : write + read
        8. a+ : append + read()
##### 2. Reading a file : read()
        - To perform the read operation on file for that we need to open file in 'r' mode & then we can perform following operations.
            1. read() : It returns entire file contant as single string.
            2. readline() : This method returns first line.
            3. readlines() : It returns list of lines.
##### 3. Writing into a file : write()
        - If you open a file in write mode we can perform following operation.
            1. write(string) : It add string into a file.
            2. writelines(List_of_Lines) : It add all list lines to a file.
##### 4. Closing a file : close()
        - After performing read / write operations on a file we need to close the file by using close method / close function, otherwise data lose     will happen.
