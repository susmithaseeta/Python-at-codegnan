```python
import math
print(math.ceil(4.25)) # It returns the Upperbound integer value
print(math.floor(4.25)) # It returns the Lowerbound integer value
print(math.factorial(5)) #  It returns the Factorial of a number
print(math.gcd(10,5,4)) # It returns the Greatest Common Devisor
print(math.lcm(10,6,4)) # It returns the Least Common Multiplier
print(math.pow(10,3)) # Pow(a, b) : It returns the Float type of a raise to b 
print(math.sqrt(9)) # It returns the Square value of a number 
print(math.log10(2)) # To find the Log10 value 
print(math.log10(4)) # To find the Log10 value 
```

```python
# who will return the top (100) in less steps [game]
import random
rolls = count = 0
print("Welcome to the Who will Reach The Top In Less Steps Game")
while True:
    input("To Throw a die, please click enter button")
    face_val = random.randint(1,6)
    
    if count + face_val <= 100:
        count += face_val
    rolls += 1
    if count == 100: 
        print(f"Your Dice value:{face_val}")
        print(f"You reached the Top in {rolls} chances")
        break
    print(f"Your Dice value:{face_val}")
    print(f"Your at {count} Position in {rolls} chances")
    print("*" * 50)
```
