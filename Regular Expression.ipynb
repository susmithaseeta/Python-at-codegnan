### Regular Expression Fuction:
 - re.match(pat,string) : It checks the weather the string is starts with pattern or not.
 - re.serach(pat, string) : It checks the entire string & it returns the first matched pattern.
 - re.findall(pat, string) : It checks the entire string & it returns list of all matched pattern.
 - re.split(pat, string) : It splits the string wherever the pattern is matched.
 - re.sub(old_pat,new_pat, string) : 
 - re.fullmatch(pat, string) : 

```python
import re
```

```python
# match
string = "The sky is blue and ocean is blue"
res = re.match("The",string)
print(res)
print(res.span()) # it returns the matched pattern position
print(res.group()) # it returns the matched result
```

```python
# match
try:
    string = "The sky is blue and ocean is blue"
    res = re.match("sky",string)
    print(res)
    print(res.span()) # it returns the matched pattern position
    print(res.group()) # it returns the matched result
except Exception as e:
    print("There is no Matched Pattern")
```

```python
# Search(pat, string) : It checks the entire string & it returns the first matched pattern.
try:
    string = "The sky is blue and ocean is blue"
    res = re.search("sky",string)
    print(res)
    print(res.span()) # it returns the matched pattern position
    print(res.group()) # it returns the matched result
except Exception as e:
    print("There is no Matched Pattern")
```

```python
# Search(pat, string) : It checks the entire string & it returns the first matched pattern.
try:
    string = "The sky is blue and ocean is blue"
    res = re.search("blue",string)
    print(res)
    print(res.span()) # it returns the matched pattern position
    print(res.group()) # it returns the matched result
except:
    print("There is no Matched Pattern")
```

```python
# findall(pat, string)
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("blue",string)
    print(res)
except:
    print("There is no Matched Pattern")
```

```python
# findall(pat, string)
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("[a,b,c]",string)
    print(res)
except:
    print("There is no Matched Pattern")
```

```python
# findall(pat, string)
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("[abco]",string)
    print(res)
except:
    print("There is no Matched Pattern")
```

```python
# Split(pat, string) 
try:
    string = "The sky is blue and ocean is blue"
    res = re.split("blue",string)
    print(res)
except:
    print("There is no Matched Pattern")
```

```python
# Split(pat, string) : It splits the string wherever the pattern is matched
try:
    string = "The sky is blue and ocean is blue"
    res = re.split("[abc]",string)
    print(res)
except:
    print("There is no Matched Pattern")
```

```python
# Sub(old_pat,new_pat, string) : It splits the string wherever the pattern is matched
try:
    string = "The sky is blue and ocean is blue"
    res = re.sub("blue", "Red",string)
    print(res)
except:
    print("There is no Matched Pattern")
```

```python
# findall(pat, string)
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("[isabce]+",string)
    print(res)
except:
    print("There is no Matched Pattern")
```

```python
# findall(pat, string)
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("[a-zA-z]+",string)
    print(res)
except:
    print("There is no Matched Pattern")
```

```python
# findall(pat, string)
try:
    string = "The sky is blue and ocean is bl9ue 2376"
    res = re.findall(r"\D+",string)
    print(res)
except:
    print("There is no Matched Pattern")
```

```python
# findall(pat, string)
try:
    string = "The sky is blue and ocean is bl9ue 2376"
    res = re.findall("[0-9]",string)
    print(res)
except:
    print("There is no Matched Pattern")
```

```python
# validate pan number
import re
try:
    email = input()
    res = re.fullmatch(r"^[a-z0-9][a-z0-9.]+\@[a-z0-9]+\.[a-z]{2,3}$",email)
    print(res)
    if res:
        print(f"{email} is valid")
    else:
        print(f"{email} is invalid")
except:
    print("There is no matched pattern")
```
